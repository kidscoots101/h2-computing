```python
# Task 1.1
import random

colours = ["blue", "red", "green"]
scores = [1, 2, 3, 4]
player_1 = []
player_2 = []

for i in range(10):
    color1 = random.randint(0, 2)
    scores1 = random.randint(0,3)
    player_1.append((colours[color1], scores1))

    color2 = random.randint(0, 2)
    scores2 = random.randint(0,3)
    player_2.append((colours[color2], scores2))

print(player_1)
print(player_2)
```

```python
# Task 1.2
def win(color1, color2):
    if color1 == color2:
        return "3"
    elif color1 == "red":
        if color2 == "green":
            return "1"
        elif color2 == "blue":
            return "2"
    elif color1 == "blue":
        if color2 == "red":
            return "1"
        elif color2 == "green":
            return "2"
    elif color1 == "green":
        if color2 == "blue":
            return "1"
        elif color2 == "red":
            return "2"
```

```python
# Task 1.3

def sortHand(hand):
    n = len(hand)
    for i in range(len(hand)):
        swapped = False
        for j in range(n-i-1):
            if hand[j][1] > hand[j+1][1]:
                hand[j], hand[j+1] = hand[j+1], hand[j]
                swapped = True
            elif hand[j][1] == hand[j+1][1]:
                if hand[j][0] < hand[j+1][0]:
                    hand[j], hand[j+1] = hand[j+1], hand[j]
        if swapped == False:
            break
    return hand
```

```python
# Task 1.4
player1_score = 0
player2_score = 0
for i in range(10):
    winner = win(player_1[i][0], player_2[i][0])
    if winner == "3":
        player1_score += player_1[i][1]
        player2_score += player_2[i][1]
    elif winner == "2":
        player1_score += player_1[i][1]
        player2_score += 2 * player_2[i][1]
    elif winner == "1":
        player1_score += 2 * player_1[i][1]
        player2_score += player_2[i][1]

print("Sorted Player 1:", sortHand(player_1))
print("Sorted Player 2:", sortHand(player_2))
print("Player 1 Score:", player1_score)
print("Player 2 Score:", player2_score)
```

```python
# Task 1.5
if player1_score > player2_score:
    print("Player 1 has won the game.")
    print("Player 2 came second.")
elif player2_score > player1_score:
    print("Player 2 has won the game.")
    print("Player 1 came second.")
else:
    print("Player 1 and 2 have tied. There is no winner.")
```
