```python
# Task 4.1
import sqlite3

conn = sqlite3.connect("company.db")

conn.execute('DROP TABLE IF EXISTS villas')
conn.execute('DROP TABLE IF EXISTS customerBookings')

conn.execute('''
    CREATE TABLE "villas" (
	"villaID"	INTEGER,
	"villaName"	TEXT,
	"country"	TEXT,
	"cost"	INTEGER,
	PRIMARY KEY("villaID")
);
''')
conn.execute('''
    CREATE TABLE "customerBookings" (
	"bookingID"	INTEGER,
	"customerID"	INTEGER,
	"villaID"	INTEGER,
	"startDate"	TEXT,
	"numOfDays"	INTEGER,
	PRIMARY KEY("bookingID"),
	FOREIGN KEY("villaID") REFERENCES "villas"("villaID")
);
''')

conn.commit()
```

```python
# Task 4.2

with open('customerBookings.txt', 'r') as customers:
    customers = customers.read().strip().split("\n")
    for customer in customers:
        customer = customer.split(",")
        
        conn.execute('''
            INSERT INTO customerBookings(bookingID, customerID, villaID, startDate, numOfDays) VALUES (?, ?, ?, ?, ?)
        ''', (customer[0], customer[1], customer[2], customer[3], customer[4]))

        conn.commit()

with open('villas.txt', 'r') as villas:
    villas = villas.read().strip().split("\n")
    for villa in villas:
        villa = villa.split(",")

        conn.execute('''
            INSERT INTO villas(villaID, villaName, country, cost) VALUES (?, ?, ?, ?)
        ''', (villa[0], villa[1], villa[2], villa[3]))

        conn.commit()

villa_data = conn.execute('''
    SELECT * FROM villas
''')

customer_data = conn.execute('''
    SELECT * FROM customerBookings
''')

print("----VILLA DATA----\n")
for villa in villa_data:
    print(villa)

print("\n----CUSTOMER DATA----\n")
for customer in customer_data:
    print(customer)
```

```python
# Task 4.3
conn.execute('DROP TABLE IF EXISTS Villa_Booking')
conn.execute('''
    CREATE TABLE "Villa_Booking" (
	"villaID"	INTEGER,
	"startDate"	TEXT,
	FOREIGN KEY("startDate") REFERENCES "customerBookings"("startDate"),
	FOREIGN KEY("villaID") REFERENCES "villas"("villaID")
);
''')

with open('customerBookings.txt', 'r') as bookings:
    bookings = bookings.read().strip().split('\n')
    for booking in bookings:
        booking = booking.split(",")
        villa_id = booking[2]
        date = booking[3]
        count = booking[4]
        if date[0] == "0":
            day = int(date[1])
        else:
            day = int(date[0:2])
        for i in range(int(count)):
            if day < 10: 
                date = "0" + str(day) + date[2:]
            else:
                date = str(day) + date[2:]
            conn.execute('''
                INSERT INTO Villa_Booking(villaID, startDate) VALUES (?, ?)
            ''', (villa_id, date))
            conn.commit()
            day += 1
```

```python
# Task 4.4

def checkAvailability():
    villaName = input("Enter villa name: ")
    month = input("Enter month: ")
    date = input("Enter date: ")
    no_of_days = int(input("Enter number of days: "))
    
    if int(date) < 10:
        startDate = "0" + date + "-" + month
    else:
        startDate = date + "-" + month
    
    villa_id = conn.execute('''
        SELECT villaID FROM villas WHERE villaName = (?)
    ''', (villaName,))
    
    villa_id = int(villa_id.fetchone()[0])
    
    villa_bookings = conn.execute('''
        SELECT startDate FROM Villa_Booking WHERE villaID = (?)
    ''', (villa_id,))
    
    dates_booked = []
    for each in villa_bookings:
        dates_booked.append(each[0])
        
    if startDate[0] == "0":
        day = int(startDate[1])
    else:
        day = int(startDate[0:2])
    
    AVAILABLE = []
    NOT_AVAILABLE = []
    for i in range(int(no_of_days)):
        if day < 10: 
            date = "0" + str(day) + startDate[2:]
        else:
            date = str(day) + startDate[2:]
        
        if date in dates_booked:
            NOT_AVAILABLE.append(date)
        elif date not in dates_booked:
            AVAILABLE.append(date)
        day += 1
    
    print("Available Dates:", AVAILABLE)
    print("Unavailable dates:", NOT_AVAILABLE)

    return

checkAvailability()
```
