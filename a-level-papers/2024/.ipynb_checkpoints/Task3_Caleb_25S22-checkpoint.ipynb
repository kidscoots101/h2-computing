```python
# Task 3.1
def heading(text):
    return "<h1>" + text + "</h1>"

def subheading(text):
    return "<h2>" + text + "</h2>"

def paragraph(text):
    return "<p>" + text + "</p>"

print(heading("My first page"))
print(subheading("My first page"))
print(paragraph("My first page"))
```

```python
# Task 3.2
def hyperlink(text):
    url, string = text.split(",")

    return f'<a href="{url}">' + string + '</a>'

print(hyperlink("https://www.cambridge.org,Click here"))
```

```python
# Task 3.3
def body(lst):
    formatted = ""
    formatted = formatted + "<html>\n<body>\n"
    for item in lst:
        element, text = item.split(": ")
        if element == "heading":
            formatted += "\t" + heading(text) + "\n"
        elif element == "subheading":
            formatted += "\t" +  subheading(text) + "\n"
        elif element == "paragraph":
            formatted += "\t" +  paragraph(text) + "\n"
        elif element == "hyperlink":
            formatted += "\t" +  hyperlink(text) + "\n"
        else:
            return False
    formatted = formatted + "</body>\n</html>"
    return formatted

lst = ["heading: This is a title", "subheading: This is a subtitle", "hyperlink: https://www.seab.gov.sg,Click"]
print(body(lst))
```

```python
# Task 3.4
def read_data(file):
    with open(file, 'r') as data:
        data = data.read().strip().split("\n")
    return body(data)
```

```python
def write_data(fileName, formatted):
    formatted = "<!doctype html>\n" + formatted

    with open(fileName + ".html", 'w') as data:
        data.write(formatted)
```

```python
file = input("File name: ")
fileName = input("Output file name: ")

formatted = read_data(file)

if formatted == False:
    print("Data cannot be converted into HTML")
else:
    write_data(fileName, formatted)
```
