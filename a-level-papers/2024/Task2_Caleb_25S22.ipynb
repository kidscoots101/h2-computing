```python
# Task 2.1
class Node:
    def __init__(self):
        self.data_value = -1
        self.right_node = -1
        self.left_node = -1
    
    def set_data(self, value):
        self.data_value = value
    def set_left(self, value):
        self.left_node = value
    def set_right(self, value):
        self.right_node = value
    
    def get_data(self):
        return self.data_value
    def get_left(self):
        return self.left_node
    def get_right(self):
        return self.right_node
```

```python
# Task 2.2
def calculate_nodes(levels):
    num = 0
    for i in range(1,levels+1):
        num += 2**(i-1)
    return f"A {levels} level number tree has {num} nodes"
```

```python
# Task 2.3
print(calculate_nodes(1))
print(calculate_nodes(3))
print(calculate_nodes(5))
print(calculate_nodes(10))
```

```python
# Task 2.4
def create_empty_tree(nodes, no_of_levels):
    total = 2**(no_of_levels-1)
    
    for i in range(total):
        nodes.append(Node())
    return nodes
```

```python
# Task 2.5
def assign_leaf_nodes(nodes, no_of_levels):
    total = 2 ** (no_of_levels - 1)

    for j in range(total):
        nodes[j].set_data(j + 1)
    return nodes

node = []
create_empty_tree(node, 3)
assign_leaf_nodes(node, 3)
for each in node:
    print(each.get_data())
```
