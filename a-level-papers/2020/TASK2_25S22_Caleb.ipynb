```python
# Task 2.1
import random

def task2_1(filename, quantity, maximum):
    with open(f'{filename}', 'w') as file:
        for i in range(quantity):
            file.write(str(random.randint(0, maximum)) + '\n')
    file.close()

task2_1('randomnumbers_Caleb_25S22.txt', 1000, 5000)
```

```python
def merge(a, b):
    i = 0
    j = 0 
    k = 0
    length = int(len(a)+len(b))
    c = [0] * length
    while i < len(a) and j < len(b):
        if a[i] > b[j]:
            c[k] = b[j]
            k += 1
            j += 1
        else:
            c[k] = a[i]
            k += 1
            i += 1
    while i < len(a):
        c[k] = a[i]
        k += 1
        i += 1
    while j < len(b):
        c[k] = b[j]
        k += 1
        j += 1
    return c

def task2_2(list_of_integers):
    if len(list_of_integers) == 1:
        return list_of_integers
    list1 = list_of_integers[0:len(list_of_integers)//2]
    list2 = list_of_integers[len(list_of_integers)//2:]
    
    list1 = task2_2(list1)
    list2 = task2_2(list2)
    
    return merge(list1, list2)
    

task2_2([56, 25, 4, 98, 0, 18, 4, 5, 7, 0]) == [0, 0, 4, 4, 5, 7, 18, 25, 56, 98]
```

```python
# Task 2.3
def task2_3(filename_in, filename_out):
    with open(filename_in, "r") as fin:
        file = fin.read()
        file = file.strip().split("\n")
        
        for i in range(len(file)):
            file[i] = int(file[i]) # NEED TO CONVERT TO INTEGER..

        file = task2_2(file)
    fin.close()
    with open(filename_out, 'w') as fout:
        for each in file:
            fout.write(str(each) + '\n')
input_file = 'randomnumbers_Caleb_25S22.txt'
output_file = 'sortednumbers_Caleb_25S22.txt'

task2_3(input_file, output_file)
```
