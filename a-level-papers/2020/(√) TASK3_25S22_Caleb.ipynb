```python
def task3_1(quantity_of_data):
    end = quantity_of_data[-2:]
    if end in ["KB", "MB", "GB", "TB"]:
        if end == "KB":
            return int(quantity_of_data[:-2]) * 1000
        elif end == "MB":
            return int(quantity_of_data[:-2]) * 1000000
        elif end == "GB":
            return int(quantity_of_data[:-2]) * 1000000000
        elif end == "TB":
            return int(quantity_of_data[:-2]) * 1000000000000

    return "invalid data"

print(task3_1("8KB"))
print(task3_1("29MB"))
print(task3_1("41GB"))
print(task3_1("38TB"))
```

```python
# Task 3.2

def task3_2(quantity_of_data):
    if quantity_of_data[-2] == "i":
        end = quantity_of_data[-3:]
    else:
        end = quantity_of_data[-2:]

    if end in ["KB", "KiB", "MB", "MiB", "GB", "GiB", "TB", "TiB"]:
        if len(end) == 2:
            return task3_1(quantity_of_data)
        else:
            if end == "KiB":
                return int(quantity_of_data[:-3]) * 2**10
            elif end == "MiB":
                return int(quantity_of_data[:-3]) * 2**20
            elif end == "GiB":
                return int(quantity_of_data[:-3]) * 2**30
            elif end == "TiB":
                return int(quantity_of_data[:-3]) * 2**40
    return "invalid data"

print(task3_2("8KB"))
print(task3_2("29MB"))
print(task3_2("41GB"))
print(task3_2("38TB"))

print(task3_2("42KiB"))
print(task3_2("2MiB"))
print(task3_2("51GiB"))
print(task3_2("67TiB"))
```

```python
# Task 3.3

def task3_3(quantity_of_data, target_unit):
    value = task3_2(quantity_of_data)

    if target_unit == "KB":
        value = value / 1000
    elif target_unit == "MB":
        value = value / 1000000
    elif target_unit == "GB":
        value = value / 1000000000
    elif target_unit == "TB":
        value = value / 1000000000000
    elif target_unit == "KiB":
        value = value / (2**10)
    elif target_unit == "MiB":
        value = value / (2**20)
    elif target_unit == "GiB":
        value = value / (2**30)
    elif target_unit == "TiB":
        value = value / (2**40)
    return value

print(task3_3("512MiB", "GiB"))
print(task3_3("25KB", "MB"))
print(task3_3("193GiB", "TiB"))
```
