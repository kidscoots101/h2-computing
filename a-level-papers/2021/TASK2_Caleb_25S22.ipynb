```python
# Task 2.1
def task2_1(filename):
    with open(filename, "r") as file: # with function automatically closes file
        content = file.read().strip().split("\n")
    arr = []
    for each in content:
        arr.append(int(each)) # conversts each item to integer
    return arr

result = task2_1('TEN.txt')
print(result)
print(len(result))
```

```python
# Task 2.2
def task2_2(list_of_integers): # insertion sort algorithm
    for i in range(len(list_of_integers)):
        curr = list_of_integers[i]
        j = i - 1
        while j >= 0 and list_of_integers[j] > curr:
            list_of_integers[j+1] = list_of_integers[j]
            j = j - 1
        list_of_integers[j + 1] = curr

    return list_of_integers

print(task2_2(task2_1('TEN.txt')))
```

```python
# Task 2.3

def task2_3(list_of_integers): # quick sort algorithm
    less = []
    more = []
    if len(list_of_integers) < 1:
        return list_of_integers

    pivot = list_of_integers[0]
    for each in list_of_integers[1:]:
        if each < pivot:
            less.append(each)
        else:
            more.append(each)
    return task2_3(less) + [pivot] + task2_3(more)

print(task2_3(task2_1('TEN.txt')))
```

```python
# import the built-in library timeit 
import timeit

# use the function from Task 1.1 to read in 1000 integers from the file THOUSAND.txt 
t1000 = task2_1('THOUSAND.txt')

t10 = task2_1('TEN.txt')
t100 = task2_1('HUNDRED.txt')

# use the timeit function to call the task2_2(li) function with the 1000 integer list, run this just once 
time1000 = timeit.timeit(lambda: task2_2(t1000), number=1)

time10 = timeit.timeit(lambda: task2_2(t10), number=1)
time100 = timeit.timeit(lambda: task2_2(t100), number=1)

# print out the time, in seconds
print(time1000)

print(time10)
print(time100)
```
