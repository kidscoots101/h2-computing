```python
# Task 1.1

def task1_1(input_value):
    if str(input_value).isnumeric() == False: # checks id input is integer, string containing integer
        return -1
    elif len(str(input_value)) != 14: # checks length
        return -1
    # input is valid
    # double all the odd numbered positions
    val = str(input_value)[::-1]
    nums = []
    for i in range(len(val)):
        index = i + 1
        if index % 2 != 0:
            nums.append(int(val[i]) * 2)
        else:
            nums.append(int(val[i]))
    total = 0
    for each in nums:
        if len(str(each)) == 1:
            total += each
        elif len(str(each)) == 2:
            total += int(str(each)[0]) + int(str(each)[1])
    
    check_digit = 10 - (total % 10)
    return check_digit

task1_1("14576567654934")
```

```python
# Task 1.2

print(task1_1("39201673847186") == 3) # string containing valid integer
print(task1_1(63728192036172) == 9) # valid integer
print(task1_1("28h10s-2sb1s91") == -1) # string containing characters that are not numbers
print(task1_1("2192") == -1) # value of incorrect length
```

```python
# Task 1.3

def task1_3(input_value):
    if str(input_value).isnumeric() == False:
        return False
    elif len(str(input_value)) != 15:
        return False

    val = str(input_value)[0:14]
    check = str(input_value)[-1]

    calc_c_digit = task1_1(val)

    return int(calc_c_digit) == int(check)

# correct IMEI numbers
print(task1_3("392016738471863"))
print(task1_3(637281920361729))
# error statements
print(not task1_3("kja1392"))
print(not task1_3("812218"))
```
