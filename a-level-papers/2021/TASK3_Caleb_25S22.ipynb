```python
# Task 3.1

class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class LinkedList:
    def __init__(self):
        self.head = None
    def insert(self, integer_value):
        new_data = Node(integer_value)
        if self.head is None:
            self.head = new_data
        else:
            new_data.next = self.head
            self.head = new_data
    def delete(self, integer_value):
        if self.head is None:
            return None
        if self.head.data == integer_value:
            self.head = self.head.next
            return True
        else:
            curr = self.head
            prev = curr
            curr = curr.next
            while curr is not None:
                if curr.data == integer_value:
                    prev.next = curr.next
                    return True
                prev = curr
                curr = curr.next
            return None
    def search(self, integer_value):
        curr = self.head
        while curr is not None:
            if curr.data == integer_value:
                return True
            curr = curr.next
        return False

    def count(self):
        curr = self.head
        count = 0
        while curr is not None:
            count += 1
            curr = curr.next
        return count
    def to_String(self):
        lst = []
        curr = self.head
        while curr is not None:
            lst.append(curr.data)
            curr = curr.next
        return str(lst)
```

```python
arr = LinkedList()
with open('Task3data.txt', 'r') as file: # with function closes file automatically
    file = file.read().strip().split('\n')
    for each in file: # iterates through each item in file
        each = int(each)
        arr.insert(each) # inserts integer into Linked List
print(arr.to_String()) # prints string containing formatted list
        
```

```python
class SortedLinkedList(LinkedList):
    def insert(self, integer_value):
        curr = self.head
        new_data = Node(integer_value)
        if self.head is None or integer_value < self.head.data:
            new_data.next = self.head
            self.head = new_data
        else:
            prev = None
            while curr is not None and integer_value >= curr.data:
                prev = curr
                curr = curr.next
            prev.next = new_data
            new_data.next = curr
```

```python
arr1 = SortedLinkedList()

with open('Task3data.txt', 'r') as file: # with function closes file automatically
    file = file.read().strip().split('\n')
    for each in file: # iterates through each item in file
        each = int(each)
        arr1.insert(each) # inserts integer into Linked List

print(arr1.to_String())
```

```python
class Queue(LinkedList):
    def enqueue(self, integer_value):
        curr = self.head
        new_data = Node(integer_value)
        if self.head is None:
            self.head = new_data
        else:
            prev = None
            while curr is not None:
                prev = curr
                curr = curr.next
            prev.next = new_data
            new_data.next = curr
    def dequeue(self):
        if self.head is None:
            return self.head
        else:
            data = self.head.data
            self.head = self.head.next
            return data
```

```python
arr2 = Queue()
with open('Task3data.txt', 'r') as file: # with function closes file automatically
    file = file.read().strip().split('\n')
    for each in file: # iterates through each item in file
        each = int(each)
        arr2.enqueue(each) # inserts integer into Linked List
print(arr2.dequeue())
print(arr2.dequeue())
print(arr2.dequeue())
print(arr2.dequeue())
print(arr2.dequeue())
```
