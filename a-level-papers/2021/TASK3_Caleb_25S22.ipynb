```python
# Task 3.1

class Node:
    def __init__(self, value):
        self.integer_value = value
        self.next = None

class LinkedList:
    def __init__(self):
        self.head = None
    def insert(self, integer_value):
        new_node = Node(integer_value)

        if self.head is None: # if linked list is empty
            self.head = new_node
        else:
            new_node.next = self.head # INSERTS AT BEGINNING OF LIST
            self.head = new_node

    def delete(self, integer_value):
        curr = self.head

        # if empty
        if curr is None:
            return None

        # value lies in head (first position)
        if curr.integer_value == integer_value:
            self.head = self.head.next

        # value lies in other postions (not first)
        while curr.next is not None:
            if curr.next.integer_value == integer_value:
                self.head.next = self.head.next.next
            curr = curr.next
        return None
    
    def search(self, integer_value):
        curr = self.head
        if curr is None:
            return False
        
        while curr is not None:
            if curr.integer_value == integer_value:
                return True
            curr = curr.next

        return False

    def count(self):
        if self.head is None: # checks if list is empty, returns 0
            return 0
        count = 0
        curr = self.head
        while curr is not None:
            count += 1 # increments for each node in Linked List
            curr = curr.next
            
        return count

    def to_String(self):
        curr = self.head
        output = ""
        while curr is not None:
            if curr.next is None: # if element is last in Linked List
                output += (str(curr.integer_value))
            else:
                output += (str(curr.integer_value) + ", ")
            curr = curr.next
            
        return "[" + output + "]"
```

```python
arr = LinkedList()
with open('Task3data.txt', 'r') as file: # with function closes file automatically
    file = file.read().strip().split('\n')
    for each in file: # iterates through each item in file
        each = int(each)
        arr.insert(each) # inserts integer into Linked List
print(arr.to_String()) # prints string containing formatted list
        
```

```python
# Task 3.2
class SortedLinkedList(LinkedList):
    def __init__(self):
        super().__init__()

    def insert(self, integer_value):
        # if list is empty
        new_node = Node(integer_value)
        if self.head is None:
            self.head = new_node
            return 

        # inserted before head
        if integer_value < self.head.integer_value:
            new_node.next = self.head
            self.head = new_node
            return
        curr = self.head

        while curr.next is not None:
            if integer_value < curr.next.integer_value:
                break
            curr = curr.next # iterates all the way till last element

        new_node.next = curr.next
        curr.next = new_node
```

```python
arr1 = SortedLinkedList()

with open('Task3data.txt', 'r') as file: # with function closes file automatically
    file = file.read().strip().split('\n')
    for each in file: # iterates through each item in file
        each = int(each)
        arr1.insert(each) # inserts integer into Linked List

print(arr1.to_String())
```

```python
# Task 3.3
class Queue(LinkedList):
    def __init__(self):
        super().__init__()
        self.array = []
    def enqueue(self, integer_value):
        self.array.append(integer_value)
        return
    def dequeue(self):
        if len(self.array) == 0:
            return None
        temp = self.array
        self.array = self.array[1:]

        return temp[0]
```

```python
arr2 = Queue()
with open('Task3data.txt', 'r') as file: # with function closes file automatically
    file = file.read().strip().split('\n')
    for each in file: # iterates through each item in file
        each = int(each)
        arr2.enqueue(each) # inserts integer into Linked List
print(arr2.dequeue())
print(arr2.dequeue())
print(arr2.dequeue())
print(arr2.dequeue())
print(arr2.dequeue())
```
