```python
# Task 3.1

def inputPuzzle():
    lst = []
    while len(lst) != 9:
        row = input("Enter a row: ")
        row = row.split(" ")
        # checks validities
        if len(row) == 9:
            count = 0
            for each in row:
                if each in "123456789":
                    count += 1
            if count == 9:
                lst.append(row)
            elif count != 0:
                print("Please enter a valid row.")
        else:
            print("Please enter a valid row.")
    return lst
```

```python
# Task 3.2
example1 = inputPuzzle()
print("Example 1:", example1)

example2 = inputPuzzle()
print("Example 2:", example2)
```

```python
# Task 3.2
def checkRow(puzzle):
    index = 0
    for row in puzzle:
        lst = []
        for each in row:
            if each in lst:
                return (index + 1) # duplicate number present
            lst.append(each)
        index = index + 1
    return -1 # puzzle is complete horizontally

check_1 = checkRow(example1)
if check_1 != -1:
    print(f"Puzzle is incomplete. Row {check_1} has duplicate values.")
else:
    print("Puzzle is complete.")

check_2 = checkRow(example2)
if check_2 != -1:
    print(f"Puzzle is incomplete. Row {check_2} has duplicate values.")
else:
    print("Puzzle is complete.")
```

```python
# Task 3.3
def checkColumn(puzzle):
    col_index = 0
    count = 0
    while count != len(puzzle):
        lst = []
        for each in puzzle:
            if each[col_index] in lst:
                return col_index + 1
            lst.append(each[col_index])
        col_index += 1
        count += 1
    return -1

col_check_1 = checkColumn(example1)
if col_check_1 != -1:
    print(f"Puzzle is incomplete. Column {col_check_1} has duplicate values.")
else:
    print("Puzzle is complete.")

col_check_2 = checkColumn(example2)
if col_check_2 != -1:
    print(f"Puzzle is incomplete. Column {col_check_2} has duplicate values.")
else:
    print("Puzzle is complete.")
```

```python
# Task 3.4
def checkSquare(puzzle):
    mini_squares = []

    lst1 = []
    lst2 = []
    lst3 = []
    for row in puzzle[0:3]: # first 3 rows
        lst1 += row[0:3]
        lst2 += row[3:6]
        lst3 += row[6:9]
    mini_squares.append(lst1)
    mini_squares.append(lst2)
    mini_squares.append(lst3)

    lst1 = []
    lst2 = []
    lst3 = []
    for row in puzzle[3:6]: # next 3 rows
        lst1 += row[0:3]
        lst2 += row[3:6]
        lst3 += row[6:9]
    mini_squares.append(lst1)
    mini_squares.append(lst2)
    mini_squares.append(lst3)

    lst1 = []
    lst2 = []
    lst3 = []
    for row in puzzle[6:9]: # last 3 rows
        lst1 += row[0:3]
        lst2 += row[3:6]
        lst3 += row[6:9]
    mini_squares.append(lst1)
    mini_squares.append(lst2)
    mini_squares.append(lst3)

    # logic to test whether mini squares are complete
    valid = True
    test_list = []
    for data in mini_squares:
        test_list = []
        for num in data:
            if num not in test_list:
                test_list.append(num)
            else:
                valid = False
    
    return valid

    
checkSquare(example2)
        
square_1 = checkSquare(example1)
if square_1:
    print('All the mini-squares in the puzzle are complete.')
else:
    print('One or more of the mini-squares in the puzzle contains duplicate values.')

square_2 = checkSquare(example2)
if square_2:
    print('All the mini-squares in the puzzle are complete.')
else:
    print('One or more of the mini-squares in the puzzle contains duplicate values.')
```
