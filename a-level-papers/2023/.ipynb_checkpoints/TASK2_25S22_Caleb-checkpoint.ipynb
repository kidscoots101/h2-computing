```python
# Task 2.1

class QueueItem:
    def __init__(self, ID, quantity): # CORRECTION: take in ID and quantity
        self.ID = ID
        self.quantity = quantity
    def getID(self):
        return self.ID
    def getQuantity(self):
        return self.quantity
```

```python
# Task 2.2
class Queue:
    def __init__(self):
        self.queuelist = [None] * 4
        self.headPointer = 0
        self.tailPointer = 0
        self.numberItems = 0
    
    # Task 2.3
    def enqueue(self, item):
        if self.numberItems == 4:
            return False
        # item = QueueItem(ID, quantity)
        # if self.headPointer == 0:
        #     self.headPointer += 1
        #     self.tailPointer += 1
        #     self.queuelist[self.tailPointer - 1] = item
        #     print(True)
        # elif self.tailPointer == 3: # queue is full
        #     print(False)
        # else:
        #     self.queuelist[self.tailPointer] = item
        #     self.tailPointer += 1
        #     print(True)
        
        self.queuelist[self.tailPointer] = item # supposed to take in item
        self.tailPointer = (self.tailPointer + 1) % 4

        self.numberItems += 1
        return True
    # Task 2.4
    def dequeue(self):
        if self.numberItems == 0: # if empty, return False
            return False

        item = self.queuelist[self.headPointer] # assigned to temp variable, to output later

        self.queuelist[self.headPointer] = None # remove first item

        self.headPointer = (self.headPointer + 1) % 4

        self.numberItems -= 1 # decrease

        return item
    # Task 2.4
    def printQueue(self):
        index = self.headPointer
        count = 0 # to keep track of where we're at
        while count < self.numberItems:
            item = self.queuelist[index]
            print("ID: ", item.getID())
            print("Quantity: ", item.getQuantity())
            count += 1
            index = (index + 1) % 4
        return

```

```python
queue = Queue()
for i in range(5):
    ID = input("Please enter an ID: ")
    quantity = int(input("Please enter a quantity: "))

    item = QueueItem(ID, quantity)

    successful = queue.enqueue(item)

    if successful: 
        print("Successfuly added into queue")
    else:
        print("Queue is full. Enqueue failed")
```

```python
for i in range(5):
    item = queue.dequeue()
    if item:
        print("ID: ", item.getID())
        print("Quantity: ", item.getQuantity())
    else:
        print("Queue is empty.")
```
