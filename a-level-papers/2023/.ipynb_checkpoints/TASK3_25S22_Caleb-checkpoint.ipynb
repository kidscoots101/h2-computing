```python
# Task 3.1

def inputPuzzle():
    lst = []
    while len(lst) != 9:
        row = input("Enter a row: ")
        row = row.split(" ")
        # checks validities
        if len(row) == 9:
            count = 0
            for each in row:
                if each in "123456789":
                    count += 1
            if count == 9:
                lst.append(row)
            elif count != 0:
                print("Please enter a valid row.")
        else:
            print("Please enter a valid row.")
    return lst

inputPuzzle()
```

```python
# Task 3.2

example1 = inputPuzzle()
print("Example 1:", example1)

example2 = inputPuzzle()
print("Example 2:", example2)
```
