```python
# Task 4.1
def readColour(file):
    with open(file, "r") as colours: # with function automatically closes file
        colours = colours.read().strip().split("\n")
    return colours
list_of_colours = readColour('imagefile.txt')

def compressData(lst):
    unique_colours = [] # ['000', '010', '011', '110', '001', '100']
    for each in lst:
        if each not in unique_colours:
            unique_colours.append(each)
    
    compress = ""
    colours_count = {}
    for i in range(len(unique_colours)): # adds all colour values to colours_count dictionary
        colours_count[unique_colours[i]] = 0

    for each in lst:
        colours_count[each] += 1

    for colour in unique_colours:
        compress += f"{colour}" + f"{colours_count[colour]}"
    
    return compress

compressed_data = compressData(list_of_colours)

def write_to_file(data):
    with open('compressedimage.txt', 'w') as text:
        text.write(data)
    print("Written compressed data to `compressedimage.txt`")

write_to_file(compressed_data)
```
