```python
import sqlite3

conn = sqlite3.connect("TESTDATABASE.db")

conn.execute('''
    DROP TABLE IF EXISTS Test
''')
conn.execute('''
    DROP TABLE IF EXISTS Class
''')

conn.execute('''
    DROP TABLE IF EXISTS Student
''')

conn.execute('''
    DROP TABLE IF EXISTS Student_Test
''')


conn.execute("""
    CREATE TABLE "Test" (
	"TestID"	TEXT,
	"MaxMarks"	INTEGER,
	PRIMARY KEY("TestID")
);
""")

conn.execute("""
    CREATE TABLE "Class" (
        "ClassName"	TEXT,
        "YearGroup"	INTEGER,
        PRIMARY KEY (ClassName)
);
""")

conn.execute("""
    CREATE TABLE "Student" (
	"StudentID"	INTEGER,
	"GivenName"	TEXT,
	"FamilyName"	TEXT,
	"ClassName"	TEXT,
	PRIMARY KEY("StudentID"),
    FOREIGN KEY (ClassName) REFERENCES Class(ClassName)
);
""")

conn.execute("""
CREATE TABLE Student_Test (
    StudentTestID INTEGER,
    StudentID INTEGER,
    TestID TEXT,
    Mark INTEGER,
    DateTest TEXT,
    PRIMARY KEY (StudentTestID),
    FOREIGN KEY (StudentID) REFERENCES Student(StudentID),
    FOREIGN KEY (TestID) REFERENCES Test(TestID)
);
""")

conn.commit()
```

```python
# Task 1.2

className = "CS3"
year = 12

studentID = 102
givenName = "Mary"
familyName = "Lim"

testID = "PG1"
maxMarks = 100

studentTestID = 123
mark = 85
date = "01-02-2023"

conn.execute("""
    INSERT INTO Class(ClassName, YearGroup) VALUES (?, ?)
""", (className, year))
conn.execute("""
    INSERT INTO Student(StudentID, GivenName, FamilyName, ClassName) VALUES (?, ?, ?, ?)
""", (studentID, givenName, familyName, className))
conn.execute("""
    INSERT INTO Test(TestID, MaxMarks) VALUES (?, ?)
""", (testID, maxMarks))
conn.execute("""
    INSERT INTO Student_Test(StudentTestID, StudentID, TestID, Mark, DateTest) VALUES (?, ?, ?, ?, ?)
""", (studentTestID, studentID, testID, mark, date))

conn.commit()
```

```python
student_test = conn.execute("""
   SELECT * FROM Student_Test 
""")
print("Student Test Data")
for data in student_test:
    print(data) # tests data has been correctly entered

class_data = conn.execute("""
   SELECT * FROM Class 
""")
print("Class Data")
for data in class_data:
    print(data)

test = conn.execute("""
   SELECT * FROM Test 
""")
print("Test Data")
for data in test:
    print(data)

student = conn.execute("""
   SELECT * FROM Student 
""")
print("Student Data")
for data in student:
    print(data)
```
