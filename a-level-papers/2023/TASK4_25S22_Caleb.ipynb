```python
# Task 4.1
def readColour(file):
    with open(file, "r") as colours: # with function automatically closes file
        colours = colours.read().strip().split("\n")
    return colours
list_of_colours = readColour('imagefile.txt')

def compressData(lst):
    compressed = ""

    current_colour = lst[0]
    count = 1

    for i in range(1, len(lst)):
        if lst[i] == current_colour:
            count += 1
        else:
            compressed += current_colour + str(count)
            current_colour = lst[i]
            count = 1

    compressed += current_colour + str(count)

    return compressed

compressed_data = compressData(list_of_colours)

def write_to_file(data):
    with open('compressedimage.txt', 'w') as text:  # with function automatically closes file
        text.write(data)
    print("Written compressed data to `compressedimage.txt`")

write_to_file(compressed_data)
```

```python
# Task 4.2
def readCompress(file):
    with open(file, 'r') as compress:
        compress = compress.read().strip()

    return compress

compressed_data = readCompress('compressedimage.txt')

def decompress(data):
    if len(data) == 4:
        return data[0:3] * int(data[3])

    return data[0:3] * int(data[3]) + decompress(data[4:])

decompressed_data = decompress(compressed_data)

def writeDecompress(data):
    with open('decompressedimage.txt', 'w') as file: # with function automatically closes file
        file.write(data)

writeDecompress(decompressed_data)
```

```python
file_name = input("Enter file name: ")
option = int(input("Choose between compressing (1), or decompressing image (2): "))

if option == 1: # compress image
    lst = readColour(file_name)
    compressed_data = compressData(lst)
    print(compressed_data)

elif option == 2:
    data = readCompress(file_name) 
    print(decompress(data))
```
