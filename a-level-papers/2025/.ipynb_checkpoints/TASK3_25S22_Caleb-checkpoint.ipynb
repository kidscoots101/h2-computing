**Task 3.1**

Write a Python program that uses SQL to create a database with the **three** tables: `User`, `Post` and `Comment`

Define the primary and foreign keys for each table. **[6]**


````python
#Task 3.1
import sqlite3

conn = sqlite3.connect('website.db')

conn.execute('DROP TABLE IF EXISTS User')
conn.execute('DROP TABLE IF EXISTS Post')
conn.execute('DROP TABLE IF EXISTS Comment')


conn.execute('''
  CREATE TABLE "User" (
	"Username"	TEXT,
	"Blocked"	INTEGER,
	PRIMARY KEY("Username")
  );
''') # creates User table

conn.execute('''
CREATE TABLE "Post" (
	"PostID"	INTEGER,
	"PostText"	TEXT,
	"DateTimePosted"	TEXT,
	"Username"	TEXT,
	PRIMARY KEY("PostID"),
	FOREIGN KEY("Username") REFERENCES "User"("Username")
);
''') # creates Post table

conn.execute('''
CREATE TABLE "Comment" (
	"CommentID"	INTEGER,
	"PostID"	INTEGER,
	"Username"	TEXT,
	"TheText"	TEXT,
	"DateTimeCommented"	TEXT,
	PRIMARY KEY("CommentID"),
	FOREIGN KEY("PostID") REFERENCES "Post"("PostID"),
	FOREIGN KEY("Username") REFERENCES "User"("Username")
);
''') # creates Comment table

conn.commit()
````

**Task 3.2**

Write a Python program that uses SQL to insert the data from each of the **three** text files into the appropriate fields in each database table. **[4]**

Run your program. **[1]**


````python
#Task 3.2
with open('user.txt', 'r') as file: # with statement automatically closes file
  file = file.read().strip().split("\n")
  for item in file:
    item = item.split(",") # splits by comma
    conn.execute('''
      INSERT INTO User VALUES(?, ?)
    ''', (item[0], item[1])) # adds data inside the table
    conn.commit()

print("---USER DATA---")
user_data = conn.execute('SELECT * FROM USER')
for each in user_data:
  print(each)
print("\n")

with open('post.txt', 'r') as file: # with statement automatically closes file
  file = file.read().strip().split("\n")
  for item in file:
    item = item.split(",")
    conn.execute('''
      INSERT INTO Post VALUES(?, ?, ?, ?)
    ''', (item[0], item[1], item[2], item[3])) # adds data inside the table
    conn.commit()

print("---POST DATA---")
post_data = conn.execute('SELECT * FROM POST')
for each in post_data:
  print(each)
print("\n")

with open('comment.txt', 'r') as file: # with statement automatically closes file
  file = file.read().strip().split("\n")
  for item in file:
    item = item.split(",")
    conn.execute('''
      INSERT INTO Comment VALUES(?, ?, ?, ?, ?)
    ''', (item[0], item[1], item[2], item[3], item[4])) # adds data inside the table
    conn.commit()

print("---COMMENT DATA---")
comment_data = conn.execute('SELECT * FROM COMMENT')
for each in comment_data:
  print(each)
````

**Task 3.3**

- outputs each returned username and comment on a new line.

Run your program. **[1]**

````python
#Task 3.3

def display(conn):
  user_blocked = conn.execute('''
    SELECT Username FROM User WHERE Blocked = 1
  ''')

  for each in user_blocked: # loops through all the users who is blocked
    username = each[0]
    comments = conn.execute('''
      SELECT TheText FROM Comment WHERE Username = ?
    ''', (username, )) # selects comment from table, where Username matches
    for comment in comments:
      print("Username: ", username)
      print("Comment: ", comment[0])
      print() # prints new line

  conn.close()

display(conn)
````
