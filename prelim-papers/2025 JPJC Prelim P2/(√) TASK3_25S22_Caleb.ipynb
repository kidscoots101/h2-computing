```python
def encrypt(code):
    output = ""
    for char in code:
        char = ord(char) - ord('a')   # Convert 'a'-'z' to 0-25
        char = (char + 3) % 26        # Shift and wrap around
        output += chr(char + ord('a'))  # Convert back to a letter
    return output

print(encrypt("abz"))
```

```python
# Task 3.2
def hashfunction(locker_id):
    hash_index = locker_id % 37
    return hash_index
```

```python
# Task 3.3
def add(locker_id, hta):
    valid = False
    while not valid:
        code_word = input("Enter a valid codeword: ")
        if code_word.islower() and code_word.isalpha():
            valid = True
        else:
            print("Please enter a valid codeword.")
    
    code_word = encrypt(code_word)

    index = hashfunction(locker_id)
    start_index = index

    while hta[index] is not None:
        index = (index + 1) % 37 # increment index, linear probing
        if index == start_index:
            print("HASH TABLE FULL")
            return hta
        
    hta[index] = [locker_id, code_word]

    return hta
```

```python
# Task 3.4
def search(hta):
    locker_ID = int(input("Please input your locker ID: "))
    code_word = input("Please input your codeword")

    code_word = encrypt(code_word)

    index = hashfunction(locker_ID) # generates index in key-value pairs
    start_index = index # to keep track of where we are coming from

    while hta[index] is not None:
        record = hta[index]

        if record[0] == locker_ID:
            if record[1] == code_word:
                print("LOCKER IS OPEN")
            else:
                print("INCORRECT CODE")
            return
        index = (index + 1) % 37
        if index == start_index:
            break
    print("LOCKER NOT FOUND")

```

```python
# Task 3.5

hta = [None] * 37 # since question says fixed size of 37

hta = add(36, hta)
hta = add(73, hta)

search(hta)
search(hta)
```
