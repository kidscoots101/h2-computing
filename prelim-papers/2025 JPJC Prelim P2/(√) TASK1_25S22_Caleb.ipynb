```python
# Task 1.1
def readData(filename):
    with open(filename, "r") as file:
        next(file)
        file = file.read().strip().split("\n")
        arr = []
        for each in file:
            each = each.split(',')
            arr.append([each[0], each[1]])
    return arr
    
arr = readData('TASK1.csv')
print(arr)
```

```python
# Task 1.2
def quicksort(treasures):
    less = []
    more = []

    if len(treasures) < 2: 
        return treasures

    pivot = treasures[0]
    for item in treasures[1:]:
        if item[0] < pivot[0]:
            less.append(item) 
        else: 
            more.append(item)
    return quicksort(less)+[pivot]+quicksort(more)

arr = quicksort(arr)
print(arr)
```

```python
# Task 1.3
def binarySearch(treasures, name):
    low = 0
    high = len(treasures) - 1
    while low <= high:
        mid = int ( (low + high) / 2 )
        if name == treasures[mid][0]:
            return treasures[mid][1]
        elif name < treasures[mid][0]:
            high = mid - 1
        else:
            low = mid + 1
    return -1

print(binarySearch(arr, "Bronze Amulet"))
```

```python
# Task 1.4

print("Welcome to Your Treasure Inventory!")
print('''
1. View Inventory
2. Sort Treausres by Name
3. Search Treasure by Name
4. Exit    
''')

done = False
while not done:
    option = int(input("Choose opton: "))
    data = readData('TASK1.csv')
    if option == 4:
        done = True
    elif option == 1:
        for each in data:
            print(f"{each[0]}" + " (Value: " + f"{each[1]}" + ")")
    elif option == 2:
        data = quicksort(data)
        for each in data:
            print(f"{each[0]}" + " (Value: " + f"{each[1]}" + ")")
    elif option == 3:
        name = str(input("Enter treasure name to search: "))
        if binarySearch(data, name) == -1:
            print("Treasure cannot be found")
        else:
            value = binarySearch(data, name)
            print(f"Found '{name}' worth {value} gold coins")


```
