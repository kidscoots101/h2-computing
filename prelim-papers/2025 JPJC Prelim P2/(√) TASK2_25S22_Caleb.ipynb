```python
# Task 2.1
class PageNode:
    def __init__(self, url):
        self.url = url
        self.next = None # CORRECTION: changed `next` to `None`

class BrowserHistory:
    def __init__(self):
        self.head = None
    
    def visit_page(self, url): # CORRECTION: had to implement Linked List using PageNode class
        new_node = PageNode(url)

        if self.head is None: # if Linked List is empty, first item of it is new_node
            self.head = new_node
        else:
            current = self.head
            while current.next is not None:
                current = current.next
            current.next = new_node
    
    
    def print_history(self): # had to iterate through Linked List and print out everything
        # print(self.head)
        current = self.head
        while current is not None:
            print(current.url)
            current = current.next
```

```python
history = BrowserHistory()

history.visit_page("google.com")
history.visit_page("youtube.com")
history.visit_page("openai.com")

history.print_history()
```

```python
# Task 2.2
class BrowserNavigator:
    def __init__(self): # we set the values to None, because they're not given by the user 
        self.current_page = None

        self.back_stack = [None] * 10
        self.forward_stack = [None] * 10

        self.back_top = -1
        self.forward_top = -1

    def go_back(self):
        if self.back_top != -1:
            self.forward_top += 1
            self.forward_stack[self.forward_top] = self.current_page

            self.current_page = self.back_stack[self.back_top]
            self.back_stack[self.back_top] = None
            self.back_top -= 1

    def go_forward(self):
        if self.forward_stack != -1:
            self.back_top += 1
            self.back_stack[self.back_top] = self.current_page

            self.current_page = self.forward_stack[self.forward_top]
            self.forward_stack[self.forward_top] = None
            self.forward_top -= 1
    def visit_new_page(self, url):
        if self.current_page is not None:
            self.back_top += 1
            self.back_stack[self.back_top] = self.current_page
        current = self.forward_top
        while current != -1:
            self.forward_stack[current] = None
            current = current - 1 # clears forward stack
        self.forward_top = -1
        self.current_page = url
        
```

```python
history = BrowserHistory()
navigator = BrowserNavigator()

navigator.visit_new_page('jpjc.coursemology.org') # MIXED UP: navigator and history functions
history.visit_page('jpjc.coursemology.org') # FORGOT: inlcude history, need to read question more in-depth

navigator.visit_new_page('jpjc.mode.edu.sg')
history.visit_page('jpjc.mode.edu.sg') 

navigator.visit_new_page('web.whatsapp.com')
history.visit_page('web.whatsapp.com')

navigator.go_back()
navigator.go_back()

navigator.go_forward()

print(navigator.current_page)
history.print_history()

```
