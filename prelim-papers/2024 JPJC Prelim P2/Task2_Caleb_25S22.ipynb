```python
# Task 2.1

import random

def createIsland(rowsize, colsize):
    first = ["*"] + ["."] * colsize
    arr = []
    arr.append(first)
    for i in range(rowsize):
        arr.append(["."] * (colsize + 1))

    return arr

arr = createIsland(9,14)

def plantCoconuts(grid):
    row = len(grid) - 1
    col = len(grid[0]) - 1

    max_coconuts = 0

    while max_coconuts < 3:
        ranRow = random.randint(0, row)
        ranCol = random.randint(0, col)

        if grid[ranRow][ranCol] == ".":
            grid[ranRow][ranCol] = "C"
            max_coconuts += 1
    return grid

grid = plantCoconuts(arr)

def display(grid):
    for each in grid:
        print(each)

# display(grid)
```
