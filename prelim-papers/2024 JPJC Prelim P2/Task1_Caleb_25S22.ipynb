```python
# Task 1.1
def CharToNum(letter):
    return ord(letter) - 65

print(CharToNum("A"))
```

```python
# Task 1.2
def encrypt(text, key):
    index = 0
    msg = ""
    for i in range(len(text)):
        if index == len(key):
            index = 0
            msg += key[index]
            index += 1
        else:
            msg += key[index]
            index += 1
    # ABLEABLEABL
    output = ""
    for j in range(len(text)):
        num = CharToNum(msg[j]) + CharToNum(text[j])
        if num >= 0 and num <= 26:
            output += chr(num + 65)
        elif num > 26:
            output += chr(num%26 + 65)
        elif num < 0:
            output += chr(abs(num + 65))
    return output

print(encrypt("HELLO WORLD", "ABLE"))
```

```python
# Task 1.3
arr = []
with open("TASK1.txt") as file:
    file = file.read()
    file = file.strip().split("\n")
    for each in file:
        text = encrypt(each, "JPJC")
        arr.append(text)
with open("encrypted.txt" , "w") as fout:
    for i in range(len(arr)):
        fout.write(arr[i] + "\n")  # forgot \n
        
    
```
