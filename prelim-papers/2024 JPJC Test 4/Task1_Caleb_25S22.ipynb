```python
# Task 1.1

def create_board():
    board = []
    for row in range(3):
        new_row = []
        for col in range(3):
            new_row.append(" ")
        board.append(new_row)
    return board

def display_board(board):
    for row in range(3):
        print(" | ".join(board[row]))
        if row < 2:
            print("---------")

def is_board_full(board):
    for row in range(3):
        for col in range(3):
            if board[row][col] == " ":
                return False
    return True

board = create_board()
display_board(board)
print(is_board_full(board))
```

```python
# Task 1.2

def check_winner(board, player):
    for row in range(3):
        if board[row][0] == board[row][1] == board[row][2] == player:
            return True

    for col in range(3):
        if board[0][col] == board[1][col] == board[2][col] == player:
            return True

    if board[0][0] == board[1][1] == board[2][2] == player:
        return True

    if board[0][2] == board[1][1] == board[2][0] == player:
        return True

    return False
```

```python
# Task 1.3
def main():
    board = create_board()
    current_player = "X"

    print("Welcome Players!")
    display_board(board)

    while True:
        print("Player", current_player + "'s turn")

        try:
            row, col = input(
                "Enter row and column (0, 1 or 2): "
            ).split()

            row = int(row)
            col = int(col)

            if row < 0 or row > 2 or col < 0 or col > 2:
                print("Row and column must be from 0 to 2.")
                continue

            if board[row][col] != " ":
                print("That cell is already occupied.")
                continue

        except ValueError:
            print("Enter two numbers separated by a space.")
            continue

        board[row][col] = current_player
        display_board(board)

        if check_winner(board, current_player):
            print("Player", current_player, "wins!")
            break

        if is_board_full(board):
            print("It's a tie!")
            break

        if current_player == "X":
            current_player = "O"
        else:
            current_player = "X"
        
```
